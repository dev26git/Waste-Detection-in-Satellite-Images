# Feature Extraction

This program extracts features of class 1 and class 0 images using the Utilities.py file.
The features are stored in a pandas dataframe and later exported as a .csv file

```python
import numpy as np
import pandas as pd
import cv2
import os
import time

# import methods of Feature Extraction
import Utilities
```

```python
# Set these folder paths

class_1_folder_path = '/Users/sinner/Desktop/Projects/SatellightSight - Waste detection using Satellite Images/All Folders/waste_only_images_all'

class_0_folder_path = '/Users/sinner/Desktop/Projects/SatellightSight - Waste detection using Satellite Images/All Folders/not_waste_images_all'
```

```python
# Creating empty DataFrame to store all features

# columns =  ['dissimilarity', 'correlation', 'homogeneity', 'contrast', 'ASM', 'energy',
#             'mean_r', 'mean_g', 'mean_b', 'std_r', 'std_g', 'std_b',
#             'skewness_r', 'skewness_g', 'skewness_b',
#             'kurtosis_r','kurtosis_g', 'kurtosis_b',
#             'Bin_0', 'Bin_1', 'Bin_2', 'Bin_3', 'Bin_4',
#             'Bin_5', 'Bin_6', 'Bin_7', 'Bin_8', 'Bin_9',
#             'Bin_10', 'Bin_11', 'Bin_12', 'Bin_13', 'Bin_14',
#             'Bin_15', 'Bin_16', 'Bin_17', 'Bin_18', 'Bin_19',
#             'Bin_20', 'Bin_21', 'Bin_22', 'Bin_23', 'Bin_24',
#             'Bin_25', 'Bin_26', 'Bin_27', 'Bin_28', 'Bin_29',
#             'class']

columns = Utilities.FEATURE_LIST
columns.append('class')

feature_df = pd.DataFrame(columns = columns)
```

```python
feature_df.columns
```

```python
def extract_features(folder_path:str, class_label:str):
    start_time = time.time()

    imgCount = 0    # Keep track of how many images scanned
    for filename in os.listdir(folder_path):

        # move to the next .jpg file
        if not filename.endswith(('.jpg', 'jpeg')):
            continue

        # import image and convert to RGB
        img = cv2.imread(os.path.join(folder_path, filename))

        # get features of image
        features = Utilities.get_features(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        features = np.ravel(features)

        # make entry into dataframe
        feature_df.loc[len(feature_df)] = np.concatenate((features, [class_label]))

        imgCount += 1

        if imgCount % 500 == 0:
            print(f"Features of {imgCount} images scanned.")

    print(f"Features of {imgCount} images extracted.")
    print(f'\nProcess completed in {time.time() - start_time} seconds.')
```

```python
# Class 1 features

extract_features(class_1_folder_path, "1")
```

```python
# Class 0 features

extract_features(class_0_folder_path, "0")
```

```python
# Converting to correct data-types

feature_df = feature_df.astype('float64')
feature_df['class'] = feature_df['class'].astype('int64')
```
