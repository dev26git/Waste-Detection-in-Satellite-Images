# Feature Extraction

This program extracts GLCM and LBP features of class 1 and class 0 images.
The features are stored in a pandas dataframe and later exported as a .csv file

```python
import numpy as np
import pandas as pd
import cv2
import os
import time

# import methods of Feature Extraction
import Utilities
```

```python
class_1_folder_path = '/Users/sinner/Desktop/waste_only_images_all'
class_0_folder_path = '/Users/sinner/Desktop/not_waste_images_all'
```

```python
# Creating empty DataFrame to store all features

columns =  ['dissimilarity', 'correlation', 'homogeneity', 'contrast', 'ASM', 'energy',
            'mean_r', 'mean_g', 'mean_b', 'std_r', 'std_g', 'std_b',
            'skewness_r', 'skewness_g', 'skewness_b',
            'kurtosis_r','kurtosis_g', 'kurtosis_b',
            'Bin_0', 'Bin_1', 'Bin_2', 'Bin_3', 'Bin_4',
            'Bin_5', 'Bin_6', 'Bin_7', 'Bin_8', 'Bin_9',
            'Bin_10', 'Bin_11', 'Bin_12', 'Bin_13', 'Bin_14',
            'Bin_15', 'Bin_16', 'Bin_17', 'Bin_18', 'Bin_19',
            'Bin_20', 'Bin_21', 'Bin_22', 'Bin_23', 'Bin_24',
            'Bin_25', 'Bin_26', 'Bin_27', 'Bin_28', 'Bin_29',
            'class']

feature_df = pd.DataFrame(columns = columns)
```

# Feature extraction of class 1 (waste)

```python
# Feature extraction of class 1 (waste)

class_1_feature_extraction_start_time = time.time()

imgCount = 0
for filename in os.listdir(class_1_folder_path):

    # move to the next .jpg file
    if not filename.endswith(('.jpg', 'jpeg')):
        continue

    # import image and convert to RGB
    img = cv2.imread(os.path.join(class_1_folder_path, filename))

    # get features of image
    features = Utilities.get_features(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    features = np.ravel(features)

    # make entry into dataframe
    feature_df.loc[len(feature_df)] = np.concatenate((features, ['1']))

    imgCount += 1
    if imgCount % 500 == 0:
        print(f"Features of {imgCount} images extracted.")

print(f"Features of {imgCount} images extracted.")
print(f'\nProcess completed in {time.time() - class_1_feature_extraction_start_time} seconds.')
```

# Feature extraction of class 0 (non-waste)

```python
# Feature extraction of class 1 (waste)

class_1_feature_extraction_start_time = time.time()

imgCount = 0
for filename in os.listdir(class_0_folder_path):

    # move to the next .jpg file
    if not filename.endswith(('.jpg', 'jpeg')):
        continue

    # import image and convert to RGB
    img = cv2.imread(os.path.join(class_0_folder_path, filename))

    # get features of image
    features = Utilities.get_features(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    features = np.ravel(features)

    # make entry into dataframe
    feature_df.loc[len(feature_df)] = np.concatenate((features, ['0']))

    imgCount += 1
    if imgCount % 500 == 0:
        print(f"Features of {imgCount} images extracted.")

print(f"Features of {imgCount} images extracted.")
print(f'\nProcess completed in {time.time() - class_1_feature_extraction_start_time} seconds.')
```

```python
feature_df = feature_df.astype('float64')
feature_df['class'] = feature_df['class'].astype('int64')
```

```python
# Save dataframe as csv
feature_df.to_csv('feature_df.csv', index=False)
```

```python
# Saving means and standard deviations for future normalization during deployment
mean_std_df = pd.DataFrame()
mean_std_df['column_names'] = feature_df.columns
mean_std_df['mean'] = feature_df.mean().values
mean_std_df['std'] = feature_df.std().values

mean_std_df.to_csv('mean_std_df.csv', index=False)
```
