# Bounding Box Extraction

We have satellite images in one folder (in .jpg format).
We have labels (yolo coordinates of bounding boxes) for each image in another folder(in .txt format).
Each bounding box marks a waste dump.

This program converts each bounding box into an image and stores all these extracted images in a new folder.
This creates our waste or class 1 dataset.

```python
import os
import cv2
import time
```

```python
# Paths to respective directories

src_folder_path = "Waste_Images_All"
labels_folder_path = "Waste_Labels_All"
class_1_folder_path = 'class_1_images'
```

```python
# Extract bounding boxes and store in class_1_folder_path

start_time = time.time()

# to assign names to extracted images
imgCount = 1

for filename in os.listdir(src_folder_path):

    # skip file if not in correct format
    if not filename.endswith(('.jpg', '.jpeg')):
        continue

    img_path = os.path.join(src_folder_path, filename)
    label_path = os.path.join(labels_folder_path, (filename[ : filename.rfind('.')] + '.txt'))

    # read image
    img = cv2.imread(img_path)

    # Get the dimensions of the image
    height, width, _ = img.shape

    # Parse the annotations from the text file
    with open(label_path, 'r') as f:
        for line in f:
            values = line.strip().split()
            class_label, x_centre, y_centre, w_norm, h_norm = map(float, values)

            # Convert the normalized coordinates to pixel coordinates
            w = round(w_norm * width)
            h = round(h_norm * height)
            x_left = round((x_centre * width) - w/2)
            x_right = round((x_centre * width) + w/2)
            y_bottom = round((y_centre * height) + h/2)
            y_top = round((y_centre * height) - h/2)

            # Correcting coordinates to be within image limits
            x_left = max(x_left, 0)
            y_top = max(y_top, 0)


            # Crop image based on annotation
            img_crop = img[y_top : y_bottom, x_left : x_right]

            # assigning a number to each cropped image
            img_number = (4 - len(str(imgCount))) * '0' + str(imgCount)

            # writing image to folder
            cv2.imwrite(os.path.join(class_1_folder_path , f'{img_number}.jpg'), img_crop)

            if imgCount % 500 == 0:
                print(f"Extracted {imgCount} images.")
            imgCount += 1

print(f"Process completed in {(time.time() - start_time)} seconds")
```
